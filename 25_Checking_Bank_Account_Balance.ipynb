#**Checking Bank Account Balance**

```python
class employees:
  def __init__(self,name,balance):
    self.name = name
    self.__ac_balance = balance   # PRIVATE

  def check_amount(self):
    print("Your Bank Balance : ",self.__ac_balance)

  def withdraw(self):
    withdraw_amount = int(input("Your withdraw amount: "))
    if withdraw_amount < self.__ac_balance:
      self.__ac_balance = self.__ac_balance - withdraw_amount
      print("Now your Balance : ",self.__ac_balance)
    else:
      print("Insufficient Balance")


obj1 = employees("Aman",50000)
obj2 = employees("Akash",80000)
```

```python
print(obj1.name)
obj1.check_amount()
obj1.withdraw()
```

#**Login Password**

```python
class students:
  def __init__(self,id,email):
    self.student_id = id
    self.student_email = email

  def login(self):
    s_student_id = input("Enter the correct student_id: ")
    s_student_email = input("Enter the correct student email: ")
    if s_student_id == self.student_id and s_student_email == self.student_email:
      print("login Successfully")
    else:
      print("Incorrect Password")
```

```python
obj1 = students("1234","abc@gmail.com")
obj1.login()
```

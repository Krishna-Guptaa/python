##**Mini Project** **:** To-Do App


```python

tasks = []

while True:
  print("\n------------------To-Do App---------------------")
  print("1. Add Task")
  print("2. View Task")
  print("3. Remove Task")
  print("4. Exit")

  choice = input("Enter you choice(1-4): ")

  if choice == "1":
    task = input("Enter the task:")
    tasks.append(task)
    print("Task Added Successfully")

  elif choice == "2":
    if len(tasks) == 0:
      print("No Task Found")
    else:
     print("\nYour Tasks: ")
     for i in range(len(tasks)):
        print(f"{i+1}.{tasks}")

  elif choice == "3":
    if len(tasks) == 0:
      print("No Task Deleted")
    else:
      for i in range(len(tasks)):
        print(f"{i+1}.{tasks}")

      delete_task = int(input("Enter the task number to delete: "))
      if delete_task > 0 or delete_task <= len(tasks):
        remove = tasks.pop(delete_task-1)
        print(f"Removed Task: {remove}")
      else:
        print("Invalid Task Number")

  elif choice == "4":
    print("Exiting To-Do App. Goodbye!")
    break

  else:
    print("Invalid Choice. Please Select 1-4.")




```
